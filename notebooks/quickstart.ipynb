### 🚀 Step 2: Run Live Cloud Extraction (`unitbytes/google-trends-scraper`)

```python
!pip install -q apify-client pandas

from getpass import getpass
import pandas as pd
from apify_client import ApifyClient

APIFY_TOKEN = getpass("Enter your Apify API Token (from https://apify.com/unitbytes/google-trends-scraper?fpr=939u3w&fp_sid=gh_gtrends_colab): ").strip()
if not APIFY_TOKEN:
    raise ValueError("Please enter a valid Apify API token to run live extraction.")

client = ApifyClient(APIFY_TOKEN)

run_input = {
    "keywords": ["OpenAI", "ChatGPT"],
    "custom_timeframe": "2023-01-01 2023-12-31",
    "common_geo": "US",
    "scrape_type": "interest_over_time",
    "compare_keywords": True,
    "gprop": "web"
}

print("Running unitbytes/google-trends-scraper...")
run = client.actor("unitbytes/google-trends-scraper").call(run_input=run_input)
items = list(client.dataset(run["defaultDatasetId"]).iterate_items())

df_live = pd.DataFrame(items)
print(f"Extracted {len(df_live)} live records!")
df_live.to_csv("google-trends-scraper-results.csv", index=False)
df_live.head(10)
```
